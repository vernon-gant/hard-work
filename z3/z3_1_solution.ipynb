```python
# 1. Roosters and piglets

roosters, piglets = Ints('roosters piglets')
 
solve(
    roosters >= 0,
    piglets >= 0,
    roosters + piglets == 11,
    2 * roosters + 4 * piglets == 30,
)
```

```python
# 2.1 Knights and knaves
# Knights always tell the truth, liars always lie. A says exactly one of
# the four is a liar, B says exactly two are, C says A and B are both
# lying, D says that if A is a knight then C is a knight. Who is who?

people = BoolVector('p', 4)
a, b, c, d = people
knight_count = Sum([If(person, 1, 0) for person in people])
 
solver = Solver()
solver.add(
    a == (knight_count == 3),
    b == (knight_count == 2),
    c == And(Not(a), Not(b)),
    d == Implies(a, c),
)
print(solver.check())
print(model = solver.model())
```

```python
# 2.2 Conference scheduling
# Five talks in five slots. Cryptography is not first, databases is
# immediately before networks, compilers and cryptography are three
# slots apart, networks is not last, compilers comes before databases.
# The fifth talk is unconstrained.

talks = IntVector("talks", 5)
crypto, databases, compilers, networks, free = talks
 
solver = Solver()
solver.add(
    Distinct(talks),
    *[And(talk >= 1, talk <= 5) for talk in talks],
    crypto != 1,
    databases + 1 == networks,
    Or(crypto - compilers == 3, crypto - compilers == -3),
    networks != 5,
    compilers < databases,
)

print(solver.check())
print(solver.model())
```

```python
# 2.3 Is the refactoring safe?
#   original:    if (list != null && list.Count == 0)
#   refactored:  if (list.Count == 0 && list != null)
# This was a tough one for me, although the solution idea is so easy

not_null, count_is_zero = Bools('not_null count_is_zero')
 
reaches_deref_original = not_null
reaches_deref_refactored = BoolVal(True)
 
solve(reaches_deref_original != reaches_deref_refactored)
solve(reaches_deref_refactored, Not(not_null))
```

```python
# 3.1 Cat, dog, mouse

cats, dogs, mice = Ints('cats dogs mice')
 
solve(
    cats > 0,
    dogs > 0,
    mice > 0,
    dogs + cats + mice == 100,
    dogs * 1500 + cats * 100 + mice * 25 == 10000,
)
```

```python
# 3.2 + 3.3 Motorcyclist

distance, acceleration, time, start_speed, end_speed = Reals('distance acceleration time start_speed end_speed')
 
equations = [
    end_speed == start_speed + acceleration * time,
    distance == start_speed * time + acceleration * time ** 2 / 2,
]
 
solve(equations + [start_speed == 30, acceleration == -8, end_speed == 0])
solve(equations + [start_speed == 0, acceleration == 6, time == Real(41)/10])
```

```python
# 3.4 Eight queens

queens = IntVector('Q', 8)
 
solver = Solver()
solver.add(
    Distinct(queens),
    *[And(queen >= 1, queen <= 8) for queen in queens],
    *[And(queens[i] - queens[j] != i - j, queens[i] - queens[j] != j - i)
      for i in range(8) for j in range(i)],
)
print(solver.check())
print(solver.model())
```
